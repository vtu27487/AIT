```python
GOAL = [1, 2, 3,
        4, 5, 6,
        7, 8, 0]

def print_board(board):
    for i in range(0, 9, 3):
        print(board[i:i + 3])
    print()

def objective(board):
    return sum(board[i] == GOAL[i] for i in range(9))

def get_neighbors(board):
    neighbors = []

    blank = board.index(0)

    row = blank // 3
    col = blank % 3

    moves = []

    if row > 0:
        moves.append(blank - 3)

    if row < 2:
        moves.append(blank + 3)

    if col > 0:
        moves.append(blank - 1)

    if col < 2:
        moves.append(blank + 1)

    for position in moves:
        new_board = board.copy()
        new_board[blank], new_board[position] = \
            new_board[position], new_board[blank]

        neighbors.append(new_board)

    return neighbors

def hill_climbing(start):
    current = start

    while True:
        print("Current board:")
        print_board(current)


        if current == GOAL:
            print("Goal reached!")
            return current

        neighbors = get_neighbors(current)


        best = max(neighbors, key=objective)

        if objective(best) <= objective(current):
            print("No better neighbor found.")
            return current

        current = best



start_board = [
    1, 2, 3,
    4, 0, 6,
    7, 5, 8
]

print("Initial board:")
print_board(start_board)

result = hill_climbing(start_board)

print("Final board:")
print_board(result)
```
