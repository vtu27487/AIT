```python
class GoalStackPlanner:

    def __init__(self):

        # Initial state
        self.state = {
            'on_floor(Monkey)',
            'at(Monkey, Window)',
            'at(Box, Window)'
        }

        # Goal
        self.goal = 'has(Monkey, Banana)'

        # Operators
        self.operators = {

            'Push(Box, Window, Center)': {
                'preconds': {
                    'on_floor(Monkey)',
                    'at(Monkey, Window)',
                    'at(Box, Window)'
                },
                'add': {
                    'at(Monkey, Center)',
                    'at(Box, Center)'
                },
                'del': {
                    'at(Monkey, Window)',
                    'at(Box, Window)'
                }
            },

            'Climb(Box)': {
                'preconds': {
                    'on_floor(Monkey)',
                    'at(Monkey, Center)',
                    'at(Box, Center)'
                },
                'add': {
                    'on_box(Monkey)'
                },
                'del': {
                    'on_floor(Monkey)'
                }
            },

            'Grasp(Banana)': {
                'preconds': {
                    'on_box(Monkey)',
                    'at(Box, Center)',
                    'at(Monkey, Center)'
                },
                'add': {
                    'has(Monkey, Banana)'
                },
                'del': set()
            }
        }

    def apply_operator(self, name):

        operator = self.operators[name]

        # Check preconditions
        if operator['preconds'].issubset(self.state):

            # Delete old facts
            self.state -= operator['del']

            # Add new facts
            self.state |= operator['add']

            print("Action:", name)
            return True

        return False

    def solve(self):

        print("Initial State:")
        print(self.state)
        print()

        # Step 1: Move box to center
        self.apply_operator('Push(Box, Window, Center)')

        # Step 2: Climb the box
        self.apply_operator('Climb(Box)')

        # Step 3: Grasp banana
        self.apply_operator('Grasp(Banana)')

        print()
        print("Final State:")
        print(self.state)
        print()

        # Check goal
        if self.goal in self.state:
            print("Goal achieved:", self.goal)
        else:
            print("Goal not achieved")


# Create planner
planner = GoalStackPlanner()

# Solve the problem
planner.solve()
```
