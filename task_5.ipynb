```python
import matplotlib.pyplot as plt

R1 = 30
R2 = 70

total = R1 + R2

P1 = R1 / total
P2 = R2 / total

print("P1 =", P1)
print("P2 =", P2)
print("P1 + P2 =", P1 + P2)

# Graph
labels = ["P1 = 30%", "P2 = 70%"]
values = [P1, P2]

plt.pie(values, labels=labels, autopct="%1.1f%%")
plt.title("P1 and P2 Probability")
plt.show()
```
